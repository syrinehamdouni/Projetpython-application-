```python
from datasets import load_dataset
import os

dataset = load_dataset("kdkd1/waste-garbage-management-dataset", split="train")

output_dir = "data/images"
os.makedirs(output_dir, exist_ok=True)

label_names = dataset.features["label"].names

for i, item in enumerate(dataset):
    image = item["image"]
    label = label_names[item["label"]]

    label_folder = os.path.join(output_dir, label)
    os.makedirs(label_folder, exist_ok=True)

    image_rgb = image.convert("RGB")

    image_rgb.save(os.path.join(label_folder, f"{i}.jpg"))

print("All images saved in folders ")
```

# Scraping

```python
import requests
from bs4 import BeautifulSoup
import os
import time


KEYWORDS = ["waste", "recycling", "garbage", "trash"]


def download_image(url, folder, count):
    try:
        response = requests.get(url, timeout=10)
        path = os.path.join(folder, f"{count}.jpg")
        with open(path, "wb") as f:
            f.write(response.content)
        print(f"Downloaded {path}")
    except Exception as e:
        print("Error downloading:", url, e)



def scrape_pixabay(keyword):
    print(f"\nScraping keyword: {keyword}")

    folder = f"data/images/{keyword}"
    os.makedirs(folder, exist_ok=True)


    base_url = f"https://pixabay.com/images/search/{keyword}/"
    response = requests.get(base_url, headers={"User-Agent":"Mozilla/5.0"})

    if response.status_code != 200:
        print("Failed to get page:", response.status_code)
        return

    soup = BeautifulSoup(response.text, "html.parser")

    
    images = soup.find_all("img")

    count = 0
    for img in images:
        src = img.get("data-src") or img.get("src")
        if not src:
            continue

        if "pixabay.com" not in src:
            continue

        count += 1
        download_image(src, folder, count)

       
        time.sleep(0.3)

    print(f"Finished keyword: {keyword}")


if __name__ == "__main__":
    for kw in KEYWORDS:
        scrape_pixabay(kw)

    print("\nScraping Completed ")
```
